Q4 — Count Frequency of Numbers
numbers = [1,2,2,3,3,3,4,4,4,4]
Create frequency dictionary.

```python
numbers = [1,2,2,3,3,3,4,4,4,4]

frequency = {}
for num in numbers:
    frequency[num] = frequency.get(num, 0) + 1
    
print("Frequency of numbers:")
for num, freq in frequency.items():
    print(f"{num}: {freq}")
```

Q5 — Word Counter
Take sentence input and count word frequency

```python

sentence = input("Enter a sentence: ")
words = sentence.split()

frequency = {}
for word in words:
    frequency[word] = frequency.get(word, 0) + 1

print("Word frequency:")
for word, freq in frequency.items():
    print(f"{word}: {freq}")
```

Q6 — Student Marks & Average
marks = {"Math":80, "English":70, "Physics":90}
Calculate total and average.

```python
marks = {"Math":80, "English":70, "Physics":90}
total_marks = sum(marks.values())
average_marks = total_marks / len(marks)
print(f"Total Marks: {total_marks}\nAverage Marks: {average_marks}")
```

Q7 — Highest Scoring Student
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}
Find highest scoring student.

```python
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}

highest_score = 0
for student, score in students.items():
    if score > highest_score:
        highest_score = score
if score == highest_score:
    print(f"Student(s) with the highest score: {student} - {score}")

# verifying the highest score using max function
print(f"Highest score:{max(students.values())}")    
```

Q9 — Mini Project: Contact Book
Add 3 contacts, search by name, print phone or Contact not found.

```python
contact_book = {}
print('1. Add Contact\n2. View Contacts\n3. Search Contact\n4. Exit')
choice = 0
while choice != 4:
    choice = int(input("Enter your choice: "))
    
    if choice == 1:
        name = input("Enter contact name: ")
        phone = input("Enter contact phone number: ")
        contact_book[name] = phone
        print(f"Contact {name} added successfully.")
        
    elif choice == 2:
        print("Contacts:")
        for name, phone in contact_book.items():
            print(f"{name}: {phone}")
            
    elif choice == 3:
        search_name = input("Enter contact name to search: ")
        if search_name.lower() in contact_book:
            print(f"{search_name}: {contact_book[search_name]}")
        else:
            print(f"Contact {search_name} not found.")
            
    elif choice == 4:
        print("Exiting the program.")
        
    else:
        print("Invalid choice. Please try again.")
```
