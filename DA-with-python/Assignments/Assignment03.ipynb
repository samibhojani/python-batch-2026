1. Write a program that accepts a list from user and print the alternate element of list.

```python
usr_list = []
choice = 0
while choice != 2:
    print("1. Add user")
    print("2. Exit")
    choice = int(input("1/2: "))
    
    if choice == 1:
        username = input("Enter username: ")
        usr_list.append(username)
        print(f"User '{username}' added.")
    elif choice == 2:
        print("Exiting...")
        print('Final user list:', usr_list)
    else:
        print("Invalid choice. Please try again.")

print('ATERNATE USER LIST:')
for user in range(0, len(usr_list), 2):
    print(f"\nUser {user}: {usr_list[user]}")
```

3. Find and display the largest number of a list without using built-in function max(). Your program
should ask the user to input values in list from keyboard.

```python
num_list = [100,20,35,21,50,200,250,125,300,1000, 450, 500]

max_number = 0
for number in num_list:
    if number > max_number:
        max_number = number
print(f"Maximum number in the list: {max_number}")
```

4. Write a program that rotates the element of a list so that the element at the first index moves to the
second index, the element in the second index moves to the third index, etc., and the element in the last
index moves to the first index.

```python
list1 = [1, 2, 3, 4, 5]
rotated = []

# Rotate the list to the right by 1 position
for i in range(len(list1), 0, -1):
    rotated.append(list1[i-1])
print(f"Original list: {list1}")
print(f"Rotated list: {rotated}")
```

5. Write a program that input a string and ask user to delete a given word from a string.

```python
sentence = input("Enter a sentence: ")
del_str = input("Enter a word to delete: ")
print(sentence)
if del_str in sentence:
    new_sentence = sentence.replace(del_str, "")
    print(f"Updated sentence: {new_sentence}")
else:
    print("Word not found in the sentence.")
```

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be
like this :
Sum of row 1 = 32
Sum of row 2 = 31
Sum of row 3 = 63

```python
matrix = [[10, 20, 30], [40, 50, 60], [70, 80, 90]]
# row wise sum
row_sum = []
for row in matrix:
    sum_row = sum(row)
    row_sum.append(sum_row)
print('Matrix:', matrix)
print(f"Row-wise sums of matrix: {row_sum}")
```
