3. Write a program in which use all the operators we can use in Python

```python
# Write your code here
num1 = 10
num2 = 20

addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
division = num1 / num2
modulus = num1 % num2
power = num1 ** num2
floor_division = num1 // num2

print(f'\n*****Arithmetic Operations*****\n1st Number: {num1}\n2nd Number: {num2}\nAddition: {addition}\nSubtraction: {subtraction}\nMultiplication: {multiplication}\nDivision: {division}\nModulus: {modulus}\nPower: {power}\nFloor Division: {floor_division}') 
print(num1 == num2)
print(num1 != num2)
print(num1 > num2)
print(num1 < num2)
num1 += num2
print(num1)
num1 -= num2
print(num1)
num1 *= num2
print(num1)
```

4. Completes the following steps of small task:

```python
# Write your code here
marks_english = 85
marks_maths = 90
marks_islamiat = 80
total_marks = 300
obtained_marks = marks_english + marks_maths + marks_islamiat
percentage = (obtained_marks / total_marks) * 100
print(f"Total Marks: {total_marks}\nObtained Marks: {obtained_marks}\nPercentage: {percentage:.2f}%")
```

6. Take the radius of a circle from a variable and calculate its Area and Circumference (use `pi = 3.14159`)

```python
# Write your code here
pi = 3.14159
radius = 5
area = pi * radius ** 2
circumference = 2 * pi * radius
print(f"Area: {area}\nCircumference: {circumference}")
```
